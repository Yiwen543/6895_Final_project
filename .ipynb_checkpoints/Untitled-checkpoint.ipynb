```python
import sys, os                                                                                                                                                                                                
sys.path.insert(0, os.path.expanduser("~/Projects/6895_Final_project"))                                                                                                                                       
os.chdir(os.path.expanduser("~/Projects/6895_Final_project"))                                                                                                                                                 
                                                                                                                                                                                                            
import time                                                                                                                                                                                                   
from llm_parser import LLMParser                             
                                                                                                                                                                                                            
TESTS = [
  ("direct — light on",      "Nova, turn on the light",            "direct_command"),                                                                                                                       
  ("direct — AC temp",       "Nova, set the AC to 24 degrees",     "direct_command"),                                                                                                                       
  ("direct — close curtain", "Nova, close the curtain",            "direct_command"),                                                                                                                       
  ("clarify — cold",         "Nova, I feel cold",                  "needs_clarification"),                                                                                                                  
  ("clarify — dark",         "Nova, it's a bit dark in here",      "needs_clarification"),                                                                                                                  
  ("clarify — boring",       "Nova, it's boring in here",          "needs_clarification"),                                                                                                                  
  ("qa — food",              "Nova, how do I make scrambled eggs", "general_qa"),                                                                                                                           
  ("invalid — greeting",     "Hello",                              "invalid"),                                                                                                                              
]                                                                                                                                                                                                             
                                                                                                                                                                                                            
llm = LLMParser()                                                                                                                                                                                             
passed = 0                                                   
total_ms = 0.0
                                                                                                                                                                                                            
for label, text, expected in TESTS:
  result, raw, ms = llm.parse_unified(text)                                                                                                                                                                 
  actual = result.get("type", "?")                                                                                                                                                                          
  total_ms += ms                                                                                                                                                                                            
  ok = actual == expected                                                                                                                                                                                   
  passed += int(ok)                                                                                                                                                                                         
  print(f"{'PASS' if ok else 'FAIL'}  [{label}]")          
  print(f"       got: {actual}  ({ms:.0f} ms)  reply: {result.get('reply','')}")                                                                                                                            
                                                                                                                                                                                                            
print(f"\n{passed}/{len(TESTS)} passed  |  avg latency: {total_ms/len(TESTS):.0f} ms")      
```
